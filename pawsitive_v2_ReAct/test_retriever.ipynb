# Test Retriever - Chroma DB

Testing the retriever with the preprocessed feline vet textbook.

```python
import sys
import os

# Add path to unified_retriever
sys.path.insert(0, os.path.join(os.path.dirname(os.getcwd()), 'pawsitive_v1_workflow/textbook_to_db'))

from langchain_huggingface import HuggingFaceEmbeddings
from langchain_experimental.open_clip import OpenCLIPEmbeddings
from langchain_chroma import Chroma
from unified_retriever import UnifiedRetriever
```

```python
# Initialize embeddings and Chroma vectorstores
persist_directory = '../chroma/Cat_Owners_Home_Veterinary_Handbook'
id_key = "doc_id"

text_embeddings = HuggingFaceEmbeddings(model_name="BAAI/bge-small-en-v1.5")
# Text collections
text_vectorstore = Chroma(
    collection_name="text_summaries_and_tables_and_image_summaries",
    embedding_function=text_embeddings,
    persist_directory=persist_directory
)

text_docstore = Chroma(
    collection_name="text_originals",
    embedding_function=text_embeddings,
    persist_directory=persist_directory
)

```

```python
# Create unified retriever
retriever = UnifiedRetriever(
    text_vectorstore, 
    text_docstore, 
    id_key=id_key
)

print("✅ Unified retriever ready")
```

```python
# Test retrieval with a sample query
test_query = "how much water should a cat drink daily"
results = retriever.retrieve_multi_modal(test_query, k=10)

print(f"📊 Retrieved {len(results)} documents for: '{test_query}'\n")

for i, doc in enumerate(results):
    modality = doc.get('modality', 'unknown')
    summary = doc.get('summary', '')[:200]
    doc_id = doc.get('doc_id', 'N/A')
    
    print(f"{i+1}. [{modality.upper()}] (ID: {doc_id})")
    print(f"   Summary: {summary}...")
    print()
```

```python
# Fetch original text by doc_id
doc_id = "4ed5de0c-040d-481f-8546-ea089e509500"

original = retriever.text_docstore._collection.get(
    ids=[doc_id], 
    include=["documents", "metadatas"]
)

original_text = original["documents"][0] if original["documents"] else None
original_metadata = original["metadatas"][0] if original["metadatas"] else None

print(f"📄 Original text for doc_id: {doc_id}\n")
print("=" * 80)
print(original_text)
print("=" * 80)
print(f"\n📋 Metadata: {original_metadata}")
```
